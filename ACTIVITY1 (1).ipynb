```python
# import the pandas library
import pandas as pd

# load the dataset you created
df = pd.read_csv("campus_energy.csv")

# display the first few rows of the data
print("Here is the raw data:")
display(df.head())

# Group the data to see total usage per building
building_totals = df.groupby("Buliding")["KWR_used"].sum()

print("\nTotal usage per building:")
print(building_totals)

# Generate a statistical summary (average, min, max) of the numeric columns
print("\nStatistical Summary:")
display(df.describe())
```
